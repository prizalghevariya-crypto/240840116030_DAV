 Practical-2: Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas

step 1 : import the pandas library

```python
import pandas as pd
```

step 2 : create sample data

```python
# Create a dictionary containing student details such as ID, Name, Age, City, and Marks.
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Amit", None, "Riya", "Ajay", "Naresh", "Jenil"],
    "Age": [20, 21, None, 20, 20, 22, 23, 28 ],
    "City": ["Surat", "Ahemdabad", "Surat", "Vadodara", "Gandhinagar", "Surat", "Ahemdabad" , "Vapi"],
    "Marks": [89, None, 75, 84, 94, 81, 90, 93]
}
```

step 3 : Inspect the database

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)

df
```

```python
# task 1 : Display the first five rows of the dataset.
# head() is useful for quickly previewing the data.

df.head()
```

```python
# task 2 : Display the last five rows of dataset.
# tail() helps check the ending records.

df.tail()
```

```python
# task 3 : Display detail information about the dataset.
# info() shows the number of rows, columns, data types, and missing values.

df.info()
```

```python
# task 4 : generate statistical information for numerical columns.
# describe() calculates count, mean, standard deviation, minimum, maximum, and quartiles.

df.describe()
```

```python
# task 5 : Display the dimenssion of dataset
# shape returns the number of rows and columns.

df.shape
```

```python
# task 6 : Display the names of all columns in the dataframe.
# columns returns the column labels.

df.columns
```

```python
# task 7 : Display the data type of easch column.
# dtypes helps identify whether columns are integers, floats, strings, etc.

df.dtypes
```

step 4 : Check missing values

```python
# task 8 : check for missing value in every column.
# isnull() identifies missing values and sum() counts them.

df.isnull().sum()
```

```python
# task 9 : replace missing values in the marks column.
# fillna() replaces missing values with the average (mean) marks.

df ["Marks"]=df["Marks"].fillna(df["Marks"].mean())
df
```

```python
# task 10 :  replace missing values in the age column.
# Missing ages are replaced with the average age.

df ["Age"]=df["Age"].fillna(df["Age"].mean())
df
```

```python
# task 11 : replace missing names with text 'unknown'.
# fillna() replaces blank names with a meaningful text value.

df ["Name"]=df["Name"].fillna("Unknown")
df
```

Step 5 : Detect duplicatr data

```python
# Task 12 : check whether each row is duplicated
# duplicated() returns True for duplicate rows and False otherwise.

df.duplicated()
```

```python
# task 13 : count the total number of duplicate row.
# sum() counts all True values returned by duplicated().

df.duplicated().sum()
```

step 6 : Remover duplicate records

```python
# task 14 : remove duplicate rows from the dataframe.
# drop_duplicates() keeps only unique records.

df = df.drop_duplicates()
df
```

step 7 : Rename column

```python
# task 15 : Rename column "Marks" to "Score".
# rename() changes the column name without affecting the data.

df.rename(columns={"Marks": "Score"}, inplace=True)
df
```

step 8 : change the data type

```python
# task 16 : convert the age column from float to integer.
# astype(int) changes the data type to integer.

df["Age"]=df["Age"].astype(int)
df.dtypes
```

step 9 : create a new column

```python
# task 17 : create a new column named "Final Score".
# Add 5 bonus marks to each student's score.

df["Final Score"]=df["Score"]+5
df
```

step 10 : convert all letter to uppercase and lowercase.

```python
# task 18 :convert all students name to uppercase
# upper() means it coverts all letters in uppercase.
df["Name"]=df["Name"].str.upper()
df
```

```python
# task 19 :convert all City name to lowercase
# upper() means it coverts all letters in lowercase.
df["City"]=df["City"].str.lower()
df
```

step 11 : Remove leading and tralling spaces from city name

```python
# str.strip() removes leading and trailing whitespace.

df["City"]=df["City"].str.strip()
df
```

step 12 : sort the dataframe based on the "Final Score" column

```python
# Highest scoring students appear first.

df.sort_values(by="Final Score", ascending=False)
```

step 13 : save the  cleaned the dataframe as a new CSV file

```python
# index=False prevents row numbers from being saved in the CSV file.

df.to_csv("Cleaned_students.csv", index=False)
print("Dataset saved successfully.")
```
