AIM : Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON ƒles using Pandas.


step 1: import the pandas library

```python
# Pandas is used to create, read, modify, and analyze tabular data.
import pandas as pd
```

step 2: Create a sample data

```python
# Create a dictionary containing student details such as ID, Name, Age, City, and Marks.
data = {
    "ID": [101, 120, 103, 104, 105, 106],
    "Name": ["Rahul", "Priya", "Rajan", "Krisha", "Ruchi", "Dhruvil"],
    "Age": [20, 21, 13, 20, 22, 20 ],
    "City": ["Surat", "Ahemdabad", "Surat", "Vadodara", "Gandhinagar", "Surat"],
    "Marks": [89, 95, 75, 84, 96, 74]
}
```

step 3 : create a dataframe

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)

df
```

step 4: Save data as a CSV file

```python
# Export the DataFrame to a CSV file without including the index column.

df.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

step 5 : Save data as a excel file

```python
# Export the DataFrame to an Excel (.xlsx) file.

df.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")
```

step 6 : Save data as a JSON file

```python
# Export the DataFrame to a JSON file using record-oriented format.

df.to_json("students.json", orient="records", index=4)
print("JSON file created successfully.")
```

step 7: raed the CSV file

```python
# Load the CSV file back into a DataFrame.
df_csv= pd.read_csv("students.csv")
df_csv
```
