Practical 6: Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie)

```python

Step 1: Import the pandas library
#Pandas is used to create, read, modify, and analyze tabular data.
import pandas as pd
import matplotlib.pyplot as plt
Step 2: Create a sample data
# Create a common dataset
Data = {
"Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
"Product": ["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"
"Region": ["North","South","East","West","North","South","East","West","North","South","East","West"],
"Sales": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
"Units_Sold": [12,30,25,14,35,30,18,38,32,21,45,40]
}

Month Product Region Sales Units_Sold
0 Jan Laptop North 120000 12
1 Feb Mobile South 135000 30
2 Mar Tablet East 150000 25
3 Apr Laptop West 145000 14
4 May Mobile North 160000 35
5 Jun Tablet South 175000 30
6 Jul Laptop East 180000 18
7 Aug Mobile West 170000 38
8 Sep Tablet North 190000 32
9 Oct Laptop South 210000 21
10 Nov Mobile East 230000 45
11 Dec Tablet West 250000 40
Next steps:
#Convert the dictionary into a Pandas DataFrame.
df = pd.DataFrame(Data)
df

Generate code with df New interactive sheet

Step 3: Create a Line Chart
#The company's sales change from January to December?
plt.figure(figsize=(10, 5))
plt.plot(Data["Month"], Data["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

Interpretation: Line Chart – Monthly Sales
The line chart shows how sales change from January to December. Sales generally increase throughout the year, from ₹1,20,000 in
January to ₹2,50,000 in December. There are small decreases in April and August.
Step 4: Create a Bar Chart
#Product generated how much total sales
plt.figure(figsize=(10, 5))
plt.bar(Data["Product"], Data["Sales"])
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()

Interpretation: Bar Chart – Total Sales by Product
The bar chart compares total sales of different products. Tablet has the highest sales, followed by Mobile, while Laptop has the lowest
sales.
Step 5: Create a Histogram

#Are the individual sales values distributed
plt.figure(figsize=(10, 5))
plt.hist(Data["Sales"], bins=10, edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

Interpretation: Histogram
The histogram shows the distribution of sales values. Most sales values are concentrated toward the higher range, showing that the
company's sales generally increased over time.
Step 6: Create a Scatter Plot
#There a relationship betweeen the number of units sold and total sales
plt.figure(figsize=(10, 5))
plt.scatter(Data["Units_Sold"], Data["Sales"])
plt.title("Sales vs Units Sold")
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.show()

Interpretation: Scatter Plot – Units Sold vs Sales
The scatter plot shows the relationship between the number of units sold and total sales. The points generally move upward, which
indicates a positive relationship. As the number of units sold increases, total sales also tend to increase.
Step 7: Create a Pie Chart
#Percentage of total sales comes from each region
plt.figure(figsize=(10, 5))
region_sales = df.groupby("Region")["Sales"].sum()
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Sales Distribution by Region")
plt.show()

Interpretation: Pie Chart – Sales by Region
The pie chart shows the percentage contribution of each region to total sales. North: 22.2% South: 24.6% East: 26.5% West: 26.7% West
contributes the highest percentage of total sales, while North contributes the lowest.
Step 8: Save the common Dataset
from google.colab import files
#Download the CSV in Google Colab
df.to_csv("sales_data.csv", index=False)
# Download CSV file
files.download("sales_data.csv")
```
