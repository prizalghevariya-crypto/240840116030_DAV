```python
import pandas as pd
```

```python
# Create a dictionary containing student details such as ID, Name, Department, City, Salary, Experience.
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Neha", "Vivek", "Pooja"],
    "Department": ["IT", "HR","IT", "Finance", "HR", "Finance", "IT", "HR"],
    "City": ["Surat", "Navsari", "Surat", "Valsad", "Surat", "Navsari", "Valsad", "Surat"],
    "Salary": [45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience": [2, 1, 5, 4, 2, 6, 8,3]
    }
```

```python
# Convert the dictionary into a Pandas DataFrame for easier data manipulation.
df = pd.DataFrame(data)

df
```

**step 1 : Display the dataset**

```python
# Display the complete dtaFrame

df
```

**step 2 : Filter data**


 Display Emplyoees from IT department.

```python
# The condition returns rows where Department is equal to "IT".

df[df["Department"]=="IT"]
```

Display emplyoees with salary greater than 50000

```python
# Display emplyoees whose salary is greater than 50000.

df[df["Salary"]>50000]
```

Display Employees from surat

```python
# The condition returns rows where City is equal to "Surat".

df[df["City"]=="Surat"]
```

Apply Multiple Conditions

```python
# Display emplyoees from It department
# having salary greater than 50000.

df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

**step 3 : sort data**

Sort salary in ascending order

```python
# Sort employees by salary from lowest to highest.

df.sort_values(by="Salary")
```

Sort salary in Descending order

```python
# Sort employees by salary from highest to kowest.

df.sort_values(by="Salary", ascending=False)
```

sort by department and salary

```python
# First sort by department.
# Within easch department , sort salaries in descending order.

df.sort_values(by=["Department", "Salary"], ascending=[True, False])
```

**step 4 : perform aggregation**

Calculate Total salary

```python
# Calculate the total salary of all employees.
df["Salary"].sum()
```

Calculate Average Salary

```python
# Calculate the average salary.
df["Salary"].mean()
```

Find Highest Salary

```python
# dilspay the maximum salary.
df["Salary"].max()
```

Find lowest salary

```python
# Diplay the minimum salary.
df["Salary"].min()
```

Count emplyoees

```python
# Count the total number of emplyoees.
df["Emp_ID"].count()
```

**step 5 : Perform GroupBy Operations**

Average salary department-wise

```python
# Group emplyoees based in Department.
# Calcutale the average salary of each department.

df.groupby("Department")["Salary"].mean()
```

Minimum salary department-wise

```python
# Calculate the minimum salary of each department.
df.groupby("Department")["Salary"].min()
```

Count salary department-wise

```python
# Calculate the total count salary of each department.
df.groupby("Department")["Salary"].count()
```

Total salary department-wise

```python
# Calculate the total salary by each department.

df.groupby("Department")["Salary"].sum()
```

Maximum Salary Department-wise

```python
# Calculate the maximum salary of each department.
df.groupby("Department")["Salary"].max()
```

Average salary department-wise

```python
# Calculate the average salary of each department.
df.groupby("Department")["Salary"].mean()
```

Group by city

```python
# calculate the average salary of each city.
df.groupby("City")["Salary"].mean()
```

**step 6 : Multiple Aggregation Functions**

```python
# apply multiple statistical functions.
# to the slary column for each department.
df.groupby("Department")["Salary"].agg(["count", "sum", "mean", "min", "max"])
```

**step 7 : save the result**

```python
# save the dataset into a CSV file.
df.to_csv("employee_data.csv", index=False)
print("Dataset saved successfully.")
```
