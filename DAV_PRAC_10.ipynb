```python

Step 1 : Install and import libraries
# Install required libraries
!pip install -q pandas plotly dash
# Import pandas for data handling
import pandas as pd
# Import plotly express for visualizations
import plotly.express as px
# Import Dash components for dashboard development
from dash import Dash, dcc, html, Input, Output
Step 2 : Load the real world Dataset

First Five Records:
Ship Mode Segment Country City State Postal

Code Region Category Sub-
Category Sales Quantity Discount Profit

0 First Class Corporate United

States Lakeland Florida 33801 South Office

Supplies Storage 563.808 4 0.2 21.1428

1 Second Class Consumer United
States
San
Francisco California 94110 West Office

Supplies Binders 36.672 2 0.2 11.4600

2 Standard
Class Corporate United

States Springfield Missouri 65807 Central Furniture Furnishings 37.300 2 0.0 17.1580
3 Standard Home United Houston Texas 77041 Central Furniture Chairs 212 058 3 0 3 -15 1470
# URL of the public Sample Superstore dataset
url = "https://raw.githubusercontent.com/johnreygoh/datasets/master/SampleSuperstore.csv"
# Load the CSV dataset into a DataFrame
original_df = pd.read_csv(url)
# Select 2000 random records
df = original_df.sample(n=2000, random_state=42)
# Reset the index
df = df.reset_index(drop=True)
# Display the first five records
print("First Five Records:")
display(df.head())
# Display dataset dimensions
print("\nDataset Shape:", df.shape)

Step 3 : Understand the dataset
# Display all column names
print("Column Names:")
print(df.columns.tolist())
# Display dataset information
print("\nDataset Information:")
df.info()
# Display statistical summary
print("\nStatistical Summary:")
display(df.describe())
240840116030

Column Names:
['Ship Mode', 'Segment', 'Country', 'City', 'State', 'Postal Code', 'Region', 'Category', 'Sub-Category', 'Sales', 'Quantity', 'Discount', 'Profit']
Dataset Information:
<class 'pandas.core.frame.DataFrame'>
RangeIndex: 2000 entries, 0 to 1999
Data columns (total 13 columns):
# Column Non-Null Count Dtype
--- ------ -------------- -----
0 Ship Mode 2000 non-null object
1 Segment 2000 non-null object
2 Country 2000 non-null object
3 City 2000 non-null object
4 State 2000 non-null object
5 Postal Code 2000 non-null int64
6 Region 2000 non-null object
7 Category 2000 non-null object
8 Sub-Category 2000 non-null object
9 Sales 2000 non-null float64
10 Quantity 2000 non-null int64
11 Discount 2000 non-null float64
12 Profit 2000 non-null float64
dtypes: float64(3), int64(2), object(8)
memory usage: 203.3+ KB
Statistical Summary:
Postal Code Sales Quantity Discount Profit
count 2000.000000 2000.000000 2000.000000 2000.000000 2000.000000
mean 55182.282500 246.613739 3.863500 0.154330 20.482826
std 32079.992711 768.588997 2.244412 0.205754 220.194133
min 1453.000000 0.836000 1.000000 0.000000 -3839.990400
25% 23666.000000 17.200000 2.000000 0.000000 1.858050
50% 55407.000000 50.988000 3.000000 0.200000 8.687750
75% 90008.000000 200.842250 5.000000 0.200000 27.384000
max 99207.000000 22638.480000 14.000000 0.800000 3177.475000
Step 4 : Data cleaning and preprocessing
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())
# Remove duplicate records
df = df.drop_duplicates()
# Remove rows with missing important values
df = df.dropna(subset=['Sales', 'Profit', 'Category', 'Region'])
# Convert sales, profit and quantity into numeric format
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
# Remove records with invalid numeric values
df = df.dropna(subset=['Sales', 'Profit', 'Quantity'])
# Display cleaned dataset
print("\nCleaned Dataset:")
display(df.head())
# Display final number of records
print("\nFinal Dataset Shape:", df.shape)

Missing Values:
Ship Mode 0
Segment 0
Country 0
City 0
State 0
Postal Code 0
Region 0
Category 0
Sub-Category 0
Sales 0
Quantity 0
Discount 0
Profit 0
dtype: int64
Duplicate Records:
2
Cleaned Dataset:
Ship Mode Segment Country City State Postal

Code Region Category Sub-
Category Sales Quantity Discount Profit

0 First Class Corporate United

States Lakeland Florida 33801 South Office

Supplies Storage 563.808 4 0.2 21.1428

1 Second Class Consumer United
States
San
Francisco California 94110 West Office

Supplies Binders 36.672 2 0.2 11.4600

2 Standard
Class Corporate United

States Springfield Missouri 65807 Central Furniture Furnishings 37.300 2 0.0 17.1580
3 Standard Home United Houston Texas 77041 Central Furniture Chairs 212 058 3 0 3 -15 1470
Step 5 : Exploratory Data Analysis
# Calculate total sales
total_sales = df['Sales'].sum()
# Calculate total profit
total_profit = df['Profit'].sum()
# Calculate total quantity sold
total_quantity = df['Quantity'].sum()
# Calculate average sales
average_sales = df['Sales'].mean()
# Display business statistics
print("TOTAL SALES:", round(total_sales, 2))
print("TOTAL PROFIT:", round(total_profit, 2))
print("TOTAL QUANTITY SOLD:", total_quantity)
print("AVERAGE SALES:", round(average_sales, 2))
# Calculate sales by category
print("\nSales by Category:")
display(df.groupby('Category')['Sales'].sum())
# Calculate sales by region
print("\nSales by Region:")
display(df.groupby('Region')['Sales'].sum())
# Calculate profit by sub-category
print("\nProfit by Sub-Category:")
display(df.groupby('Sub-Category')['Profit'].sum())

TOTAL SALES: 493182.51
TOTAL PROFIT: 40952.02
TOTAL QUANTITY SOLD: 7720
AVERAGE SALES: 246.84
Sales by Category:
Sales
Category
Furniture 150854.6645
Office Supplies 152725.8140
Technology 189602.0350
dtype: float64
Sales by Region:
Sales
Region
Central 88554.3960
East 140466.6960
South 111031.8930
West 153129.5285
dtype: float64
Profit by Sub-Category:
Profit
Sub-Category
Accessories 8226.9865
Appliances 3536.6474
Art 1506.7986
Binders 9456.5234
Bookcases -1607.6044
Chairs 3905.6126
Copiers 4240.9164
Envelopes 1327.6136
Fasteners 151.4073
Furnishings 2184.2611
Labels 1271.6502
Machines -6162.6396
Paper 6661.4655
Phones 10674.0567
Storage 5250.6937
Supplies -1496.1178
Tables -8176.2535
dtype: float64
Step 6 : Create data visualizations
import plotly.io as pio
# Set renderer to colab to display plots inline
pio.renderers.default = 'colab'
# Calculate total sales for each category
category_sales = df.groupby(
'Category', as_index=False
)['Sales'].sum()
# Create an interactive bar chart
fig1 = px.bar(
category_sales,
x='Category',
y='Sales',
color='Category',
title='Sales by Category',
text_auto='.2s'
)
# Adjust chart layout
fig1.update_layout(
yaxis_title="Sales (₹)",
template='plotly_white',
height=400
)
# Display the chart
fig1.show()
# Calculate total sales for each region

150k 150k

190k

Furniture Office Supplies Technology 0
50k
100k
150k
Sales by Category

Category

Sales (₹)
region_sales = df.groupby(
'Region', as_index=False
)['Sales'].sum()
# Create an interactive pie chart
fig2 = px.pie(
region_sales,
names='Region',
values='Sales',
title='Sales Distribution by Region',
hole=0.35
)
# Display profit values in rupees on hover
fig2.update_traces(
hovertemplate='%{label}<br>Profit: ₹%{value:,.2f}<extra></extra>'
)
# Display the chart
fig2.show()
# Calculate total profit for each sub-category
subcategory_profit = df.groupby(
'Sub-Category', as_index=False
)['Profit'].sum()
# Sort sub-categories according to profit
subcategory_profit = subcategory_profit.sort_values(
by='Profit', ascending=False
)
# Create an interactive horizontal bar chart
fig3 = px.bar(
subcategory_profit,
x='Profit',
y='Sub-Category',
orientation='h',
color='Profit',
title='Profit by Sub-Category'
)
# Display profit values in rupees
fig3.update_layout(
xaxis_title="Profit (₹)",
template="plotly_white"
)
# Display the chart
fig3.show()

Step 7 : Design the Dashboard Layout
# Create the Dash application
app = Dash(__name__)
# Define the dashboard layout
app.layout = html.Div([
# Main dashboard heading
html.H1(
"SUPERSTORE SALES DASHBOARD",
style={
'textAlign': 'center',
'color': 'white'
}
),
# Dashboard subtitle
html.H3(
"Retail Business Performance Analysis",
style={
'textAlign': 'center',
'color': '#00D4FF'
}
),
html.Br(),
# Region filter heading
html.Label(
"Select Region",
style={'color': 'white'}
),
# Region dropdown
dcc.Dropdown(
id='region-filter',
options=[
{'label': 'All Regions', 'value': 'All'}
] + [
{'label': r, 'value': r}
for r in sorted(df['Region'].dropna().unique())
],
value='All',
clearable=False
),
html.Br(),
# Category filter heading
html.Label(
"Select Category",
style={'color': 'white'}
),
# Category dropdown
dcc.Dropdown(
id='category-filter',
options=[
{'label': 'All Categories', 'value': 'All'}
] + [
{'label': c, 'value': c}
for c in sorted(df['Category'].dropna().unique())
],
value='All',
clearable=False
),
html.Br(),
# KPI cards section
html.Div([
# Total sales card
html.Div([
html.H4("Total Sales"),
html.H2(id='sales-kpi')
], style={
'backgroundColor': '#243B55',
'color': 'white',
'padding': '15px',
'width': '30%',
'textAlign': 'center',
'borderRadius': '10px'
}),
# Total profit card
html.Div([
html.H4("Total Profit"),
html.H2(id='profit-kpi')
], style={
'backgroundColor': '#243B55',
'color': 'white',
'padding': '15px',
'width': '30%',
'textAlign': 'center',
'borderRadius': '10px'
}),
# Total quantity card
html.Div([
html.H4("Total Quantity"),
html.H2(id='quantity-kpi')
], style={
'backgroundColor': '#243B55',
'color': 'white',
'padding': '15px',
'width': '30%',

'textAlign': 'center',
'borderRadius': '10px'
})
], style={
'display': 'flex',
'justifyContent': 'space-around',
'gap': '15px'
}),
html.Br(),
# Sales by category chart
dcc.Graph(id='category-chart'),
# Sales by region chart
dcc.Graph(id='region-chart'),
# Profit by sub-category chart
dcc.Graph(id='profit-chart')
], style={
'backgroundColor': '#141E30',
'padding': '20px'
})
Step 8 : Add Interactivity
# Define callback to update dashboard components
@app.callback(
# Output components to be updated
Output('sales-kpi', 'children'),
Output('profit-kpi', 'children'),
Output('quantity-kpi', 'children'),
Output('category-chart', 'figure'),
Output('region-chart', 'figure'),
Output('profit-chart', 'figure'),
# Input values from dropdown filters
Input('region-filter', 'value'),
Input('category-filter', 'value')
)
# Function to update dashboard
def update_dashboard(region, category):
# Create a copy of the original dataset
filtered = df.copy()
# Apply region filter
if region != 'All':
filtered = filtered[
filtered['Region'] == region
]
# Apply category filter
if category != 'All':
filtered = filtered[
filtered['Category'] == category
]
# Calculate KPI values
total_sales = filtered['Sales'].sum()
total_profit = filtered['Profit'].sum()
total_quantity = filtered['Quantity'].sum()
# Group sales according to category
c = filtered.groupby(
'Category', as_index=False
)['Sales'].sum()
# Create category sales chart
fig1 = px.bar(
c,
x='Category',
y='Sales',
color='Category',
title='Sales by Category'
)
# Group sales according to region
r = filtered.groupby(
'Region', as_index=False
)['Sales'].sum()
# Create regional sales pie chart
fig2 = px.pie(
r,
names='Region',
values='Sales',
title='Sales by Region'
)
# Group profit according to sub-category
s = filtered.groupby(
'Sub-Category', as_index=False
)['Profit'].sum()
# Create sub-category profit chart
fig3 = px.bar(
s,
x='Profit',
y='Sub-Category',
orientation='h',
color='Profit',
title='Profit by Sub-Category'
)
# Return KPI values

return (
f"₹{total_sales:,.2f}",
f"₹{total_profit:,.2f}",
total_quantity,
fig1,
fig2,
fig3
)
Step 9 : Run the Interactive Dashboard

Dash app running on:
Warning: This function may stop working due to changes in browser security.
Try `serve_kernel_port_as_iframe` instead.
# Run the dashboard inside Google Colab
app.run(jupyter_mode="external", debug=False)

Step 10 : Generate actionable business insights
# Find the category with highest sales
best_category = df.groupby('Category')['Sales'].sum().idxmax()
# Find the region with highest sales
best_region = df.groupby('Region')['Sales'].sum().idxmax()
# Find the sub-category with lowest profit
low_profit_subcategory = df.groupby(
'Sub-Category'
)['Profit'].sum().idxmin()
# Find the sub-category with highest profit
high_profit_subcategory = df.groupby(
'Sub-Category'
)['Profit'].sum().idxmax()
# Display business insights
print("ACTIONABLE BUSINESS INSIGHTS")
print("\n1. Highest Sales Category:", best_category)
print("2. Highest Sales Region:", best_region)
print("3. Highest Profit Sub-Category:", high_profit_subcategory)
print("4. Lowest Profit Sub-Category:", low_profit_subcategory)
print("5. Total Sales:", round(df['Sales'].sum(), 2))
print("6. Total Profit:", round(df['Profit'].sum(), 2))
ACTIONABLE BUSINESS INSIGHTS
1. Highest Sales Category: Technology
2. Highest Sales Region: West
3. Highest Profit Sub-Category: Phones
4. Lowest Profit Sub-Category: Tables
5. Total Sales: 493182.51
6. Total Profit: 40952.02
Step 11 : Ethical Considerations
# Display ethical considerations
print("ETHICAL CONSIDERATIONS")
print("\n1. Data Privacy:")
print("Protect customer information and personal details.")
print("\n2. Data Security:")
print("Prevent unauthorized access to business data.")
print("\n3. Data Accuracy:")
print("Use correct and reliable data for analysis.")
print("\n4. Transparency:")
print("Clearly explain the data collection and analysis process.")
print("\n5. Fairness:")
print("Avoid biased decisions based on incomplete data.")
ETHICAL CONSIDERATIONS
1. Data Privacy:
Protect customer information and personal details.
2. Data Security:
Prevent unauthorized access to business data.
3. Data Accuracy:
Use correct and reliable data for analysis.
4. Transparency:
Clearly explain the data collection and analysis process.
5. Fairness:
Avoid biased decisions based on incomplete data.
```
